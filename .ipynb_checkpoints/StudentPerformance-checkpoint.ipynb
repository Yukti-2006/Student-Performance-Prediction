```python
import pandas as pd
data=pd.read_csv("C:/StudentPerformanceFactors.csv")
data.head()
```

```python
data.info()
```

```python
print(data.isnull().sum())
```

```python
data.drop(['Parental_Education_Level','Distance_from_Home','Extracurricular_Activities','School_Type','Teacher_Quality'],axis=1,inplace=True)
```

```python
print(data.isnull().sum())
```

```python
data['Parental_Involvement'] = data['Parental_Involvement'].replace({'Low': 0, 'Medium': 1,'High':2})
data['Access_to_Resources']= data['Access_to_Resources'].replace({'Low': 0, 'Medium': 1,'High':2})
data['Motivation_Level']= data['Motivation_Level'].replace({'Low': 0, 'Medium': 1,'High':2})
data['Family_Income']= data['Family_Income'].replace({'Low': 0, 'Medium': 1,'High':2})
data['Internet_Access']= data['Internet_Access'].replace({'No': 0, 'Yes': 1})
data['Gender']= data['Gender'].replace({'Male': 0, 'Female': 1})
data['Learning_Disabilities']= data['Learning_Disabilities'].replace({'No': 0, 'Yes': 1})
data['Peer_Influence']= data['Peer_Influence'].replace({'Negative': 0, 'Neutral': 1,'Positive':2})
data.head()
```

```python
print(data.isnull().sum())
```

```python
x=data[['Hours_Studied','Attendance','Parental_Involvement','Gender','Access_to_Resources','Tutoring_Sessions','Sleep_Hours','Previous_Scores','Motivation_Level','Internet_Access','Family_Income','Peer_Influence','Physical_Activity','Learning_Disabilities']]
y=data['Exam_Score']
#print(x,y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state= 42 )
```

```python
from sklearn.ensemble import RandomForestClassifier
model= RandomForestClassifier()
model.fit(x_train,y_train)
```

```python
from sklearn.linear_model import LinearRegression
model= LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred= model.predict(x_test)
print(y_pred)
```

```python
from math import sqrt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print('Mean absolute error:%.2f' % mean_absolute_error(y_test, y_pred))
print('Mean squared error:%.2f' % mean_squared_error(y_test, y_pred))
print('Root mean absolute error:%.2f' % sqrt( mean_squared_error(y_test, y_pred)))
print('R2-score:%.2f'% r2_score(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt

# Generate predictions
y_all_pred = model.predict(x)

plt.figure(figsize=(8, 6))

# Plot Actual vs Predicted
plt.scatter(y, y_all_pred, color='purple', alpha=0.7)

# Plot the reference line (Perfect prediction line)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linestyle='--', linewidth=2)

plt.title('Actual vs. Predicted Values')
plt.xlabel('Actual Y')
plt.ylabel('Predicted Y')
plt.grid(True)
plt.show()
```
